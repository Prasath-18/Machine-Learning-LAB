```python
import csv


attr_values = [
    ['>=9', '>=8'],
    ['Yes', 'No'],
    ['Excellent', 'Good'],
    ['Good', 'Bad'],
    ['Fast', 'Slow'],
    ['Yes', 'No']
]


num_attrs = len(attr_values)


print("\nThe Most General Hypothesis:")
print(['?'] * num_attrs)

print("\nThe Most Specific Hypothesis:")
print(['0'] * num_attrs)


training_data = []

print("\nTraining Dataset:\n")

with open("job_offer_dataset.csv", "r") as csvFile:
    reader = csv.reader(csvFile)

    
    next(reader)

    for row in reader:
        training_data.append(row)
        print(row)


hypothesis = ['0'] * num_attrs

print("\nInitial Hypothesis:")
print(hypothesis)


for row in training_data:
    if row[num_attrs] == "Yes":
        for j in range(num_attrs):
            hypothesis[j] = row[j]
        break


print("\nFind-S: Finding Maximally Specific Hypothesis\n")

for i in range(len(training_data)):
    if training_data[i][num_attrs] == "Yes":
        for j in range(num_attrs):
            if training_data[i][j] != hypothesis[j]:
                hypothesis[j] = '?'

    print(f"Training Instance {i + 1}: {hypothesis}")


print("\nFinal Maximally Specific Hypothesis:\n")
print(hypothesis)
```

```python
import csv


data = []

with open("job_offer_dataset.csv", "r") as csvfile:
    reader = csv.reader(csvfile)
    next(reader)  
    for row in reader:
        data.append(row)


num_attributes = len(data[0]) - 1


S = ['0'] * num_attributes
G = ['?'] * num_attributes

print("Training Data:\n")
for row in data:
    print(row)

print("\nInitial Specific Hypothesis (S):")
print(S)

print("\nInitial General Hypothesis (G):")
print(G)


for i, row in enumerate(data):

    if row[-1] == "Yes":      

        
        if S[0] == '0':
            S = row[:-1].copy()
        else:
            for j in range(num_attributes):
                if row[j] != S[j]:
                    S[j] = '?'

        
        for j in range(num_attributes):
            if G[j] != '?' and G[j] != S[j]:
                G[j] = '?'

    else:                     

        for j in range(num_attributes):

            if S[j] != '?' and row[j] != S[j]:
                G[j] = S[j]

    print("\nAfter Training Instance", i + 1)
    print("S =", S)
    print("G =", G)

print("\nFinal Specific Hypothesis:")
print(S)

print("\nFinal General Hypothesis:")
print(G)
```
