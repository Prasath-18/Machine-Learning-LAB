```python

import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def sigmoid_derivative(x):
    return x * (1 - x)

x1 = 0
x2 = 1

inputs = np.array([[x1, x2]])
expected_output = np.array([[1]])

lr = 0.8

w13 = -0.2
w23 =  0.2

w14 =  0.4
w24 = -0.3

w35 =  0.2
w45 = -0.3

theta3 =  0.4
theta4 =  0.1
theta5 = -0.3
epoch = 0

while True:

    epoch += 1
    net3 = (x1 * w13) + (x2 * w23) - theta3

    x3 = sigmoid(net3)
    net4 = (x1 * w14) + (x2 * w24) - theta4

    x4 = sigmoid(net4)

    net5 = (x3 * w35) + (x4 * w45) - theta5

    x5 = sigmoid(net5)

    error = expected_output - x5

    if abs(error[0][0]) < 0.001:
        break
    delta5 = error * sigmoid_derivative(x5)


    error3 = delta5 * w35
    delta3 = error3 * sigmoid_derivative(x3)


    error4 = delta5 * w45
    delta4 = error4 * sigmoid_derivative(x4)

    w35 = w35 + lr * x3 * delta5[0][0]

    w45 = w45 + lr * x4 * delta5[0][0]
    w13 = w13 + lr * x1 * delta3[0][0]
    w23 = w23 + lr * x2 * delta3[0][0]

    w14 = w14 + lr * x1 * delta4[0][0]
    w24 = w24 + lr * x2 * delta4[0][0]

    theta3 = theta3 - lr * delta3[0][0]

    theta4 = theta4 - lr * delta4[0][0]

    theta5 = theta5 - lr * delta5[0][0]


print("Final Weights:")

print("w13 =", w13)
print("w23 =", w23)

print("w14 =", w14)
print("w24 =", w24)

print("w35 =", w35)
print("w45 =", w45)

print("\nFinal Bias Values:")
print("theta3 =", theta3)
print("theta4 =", theta4)
print("theta5 =", theta5)

print("\nFinal Output:")
print("x3 =", x3)
print("x4 =", x4)
print("x5 =", x5)

print("\nExpected Output =", expected_output[0][0])
print("Predicted Output =", x5)
print("\nNumber of Epochs =", epoch)

```

```python
from keras.models import Sequential
from keras.layers import Dense, Input
from keras.optimizers import Adam
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_blobs

X, y = make_blobs(
    n_samples=100,
    centers=2,
    n_features=2,
    random_state=0
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=0
)


model = Sequential([
    Input(shape=(2,)),
    Dense(12, activation='relu'),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    loss='binary_crossentropy',
    optimizer=Adam(learning_rate=0.01),
    metrics=['accuracy']
)

print("Initial Training")

model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=10,
    verbose=0
)


scores = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Initial Accuracy: %.2f%%" % (scores[1] * 100))
print("\nFine-Tuning")

model.compile(
    loss='binary_crossentropy',
    optimizer=Adam(learning_rate=0.001),
    metrics=['accuracy']
)

model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=10,
    verbose=0
)

scores = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Fine-Tuned Accuracy: %.2f%%" % (scores[1] * 100))
```
